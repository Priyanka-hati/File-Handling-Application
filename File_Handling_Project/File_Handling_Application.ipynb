```python
import os
import shutil

print("Modules imported successfully.")
```

```python
# ============================================================
# PARENT CLASS: FileManager
# ============================================================

class FileManager:

    # --------------------------------------------------------
    # CONSTRUCTOR
    # --------------------------------------------------------
    def __init__(self):
        self.app_name = "FILE HANDLING APPLICATION"


    # ========================================================
    # OPTION 1: CREATE / WRITE FILE
    # ========================================================
    def create_file(self):

        try:
            filename = input("Enter file name: ").strip()
            content = input("Enter content: ")

            with open(filename, "w") as file:
                file.write(content + "\n")

            # Get the complete location of the file
            file_path = os.path.abspath(filename)

            print("File created/written successfully.")
            print("File stored at:", file_path)

        except PermissionError:
            print("Error: Permission denied.")

        except OSError as e:
            print("Error:", e)


    # ========================================================
    # OPTION 2: READ FILE
    # ========================================================
    def read_file(self):

        try:
            filename = input("Enter file name: ").strip()

            with open(filename, "r") as file:
                content = file.read()

            print("\n----- File Content -----")
            print(content, end="")
            print("\n------------------------")

            file_path = os.path.abspath(filename)
            print("File location:", file_path)

        except FileNotFoundError:
            print("Error: File not found.")

        except PermissionError:
            print("Error: Permission denied.")

        except OSError as e:
            print("Error:", e)


    # ========================================================
    # OPTION 3: APPEND TO FILE
    # ========================================================
    def append_file(self):

        try:
            filename = input("Enter file name: ").strip()
            content = input("Enter content to append: ")

            with open(filename, "a") as file:
                file.write(content + "\n")

            file_path = os.path.abspath(filename)

            print("Content appended successfully.")
            print("File location:", file_path)

        except FileNotFoundError:
            print("Error: File not found.")

        except PermissionError:
            print("Error: Permission denied.")

        except OSError as e:
            print("Error:", e)


    # ========================================================
    # OPTION 4: SEARCH IN FILE
    # ========================================================
    def search_file(self):

        try:
            filename = input("Enter file name: ").strip()
            word = input("Enter word/phrase to search: ").strip()

            with open(filename, "r") as file:
                content = file.read()

            if word.lower() in content.lower():
                print("Word/phrase found in the file.")
            else:
                print("Word/phrase not found in the file.")

            file_path = os.path.abspath(filename)
            print("File location:", file_path)

        except FileNotFoundError:
            print("Error: File not found.")

        except PermissionError:
            print("Error: Permission denied.")

        except OSError as e:
            print("Error:", e)


    # ========================================================
    # OPTION 5: COUNT LINES
    # ========================================================
    def count_lines(self):

        try:
            filename = input("Enter file name: ").strip()

            with open(filename, "r") as file:
                lines = file.readlines()

            print("Number of lines:", len(lines))

            file_path = os.path.abspath(filename)
            print("File location:", file_path)

        except FileNotFoundError:
            print("Error: File not found.")

        except PermissionError:
            print("Error: Permission denied.")

        except OSError as e:
            print("Error:", e)


    # ========================================================
    # OPTION 6: COPY FILE
    # ========================================================
    def copy_file(self):

        try:
            source = input("Enter source file name: ").strip()
            destination = input("Enter destination file name: ").strip()

            if os.path.exists(destination):
                raise FileExistsError

            shutil.copy2(source, destination)

            destination_path = os.path.abspath(destination)

            print("File copied successfully.")
            print("Copied file stored at:", destination_path)

        except FileNotFoundError:
            print("Error: Source file not found.")

        except FileExistsError:
            print("Error: Destination file already exists.")

        except PermissionError:
            print("Error: Permission denied.")

        except OSError as e:
            print("Error:", e)


    # ========================================================
    # OPTION 7: RENAME FILE
    # ========================================================
    def rename_file(self):

        try:
            old_name = input("Enter current file name: ").strip()
            new_name = input("Enter new file name: ").strip()

            if not os.path.exists(old_name):
                raise FileNotFoundError

            if os.path.exists(new_name):
                raise FileExistsError

            os.rename(old_name, new_name)

            new_path = os.path.abspath(new_name)

            print("File renamed successfully.")
            print("File stored at:", new_path)

        except FileNotFoundError:
            print("Error: Current file not found.")

        except FileExistsError:
            print("Error: New file name already exists.")

        except PermissionError:
            print("Error: Permission denied.")

        except OSError as e:
            print("Error:", e)


    # ========================================================
    # OPTION 8: DELETE FILE
    # ========================================================
    def delete_file(self):

        try:
            filename = input("Enter file name: ").strip()

            if not os.path.exists(filename):
                raise FileNotFoundError

            file_path = os.path.abspath(filename)

            confirmation = input(
                "Are you sure you want to delete it? (yes/no): "
            ).strip().lower()

            if confirmation == "yes":
                os.remove(filename)

                print("File deleted successfully.")
                print("Deleted file was located at:", file_path)

            else:
                print("Delete operation cancelled.")

        except FileNotFoundError:
            print("Error: File not found.")

        except PermissionError:
            print("Error: Permission denied.")

        except OSError as e:
            print("Error:", e)
```

```python
# ============================================================
# CHILD CLASS: FileHandlingApp
# ============================================================

class FileHandlingApp(FileManager):

    # --------------------------------------------------------
    # CONSTRUCTOR
    # --------------------------------------------------------
    def __init__(self):
        super().__init__()


    # ========================================================
    # MAIN MENU
    # ========================================================
    def main(self):

        while True:

            print("\n==========================================")
            print("       FILE HANDLING APPLICATION")
            print("==========================================")
            print("1. Create / Write File")
            print("2. Read File")
            print("3. Append to File")
            print("4. Search in File")
            print("5. Count Lines")
            print("6. Copy File")
            print("7. Rename File")
            print("8. Delete File")
            print("9. Exit")
            print("==========================================")

            choice = input("Enter your choice: ").strip()


            # =================================================
            # CHOICE 1: CREATE / WRITE
            # =================================================
            if choice == "1":

                print("\n--- CREATE / WRITE FILE ---")
                self.create_file()


            # =================================================
            # CHOICE 2: READ
            # =================================================
            elif choice == "2":

                print("\n--- READ FILE ---")
                self.read_file()


            # =================================================
            # CHOICE 3: APPEND
            # =================================================
            elif choice == "3":

                print("\n--- APPEND TO FILE ---")
                self.append_file()


            # =================================================
            # CHOICE 4: SEARCH
            # =================================================
            elif choice == "4":

                print("\n--- SEARCH IN FILE ---")
                self.search_file()


            # =================================================
            # CHOICE 5: COUNT LINES
            # =================================================
            elif choice == "5":

                print("\n--- COUNT LINES ---")
                self.count_lines()


            # =================================================
            # CHOICE 6: COPY
            # =================================================
            elif choice == "6":

                print("\n--- COPY FILE ---")
                self.copy_file()


            # =================================================
            # CHOICE 7: RENAME
            # =================================================
            elif choice == "7":

                print("\n--- RENAME FILE ---")
                self.rename_file()


            # =================================================
            # CHOICE 8: DELETE
            # =================================================
            elif choice == "8":

                print("\n--- DELETE FILE ---")
                self.delete_file()


            # =================================================
            # CHOICE 9: EXIT
            # =================================================
            elif choice == "9":

                print("\nThank you for using the application.")
                break


            # =================================================
            # INVALID CHOICE
            # =================================================
            else:

                print("\nInvalid choice.")
                print("Please enter a number from 1 to 9.")
```

```python
# ============================================================
# START FILE HANDLING APPLICATION
# ============================================================

app = FileHandlingApp()
app.main()
```
